#### Imports

```python
# Model Imports
from ultralytics import YOLO

# Additional Imports
import os
from pathlib import Path
from zoneinfo import ZoneInfo
from datetime import datetime 
```

#### Path Configurations

```python
# Path Configuration
current_working_directory = Path.cwd()
base_project_directory = current_working_directory.parent.parent.parent

# Base Model and Project Path Configuration
base_model_name = "yolo26l"
current_date = datetime.now(ZoneInfo("Europe/Berlin")).strftime("%d-%m-%Y_%H-%M")
project_name = f"{current_date}_{base_model_name}"

# Dataset Path Configuration
data_directory = current_working_directory.parent.parent.parent / "data" / "detection"
dataset_name = "football-players-detection.v11i.yolov8"
full_dataset_path = os.path.join(data_directory, dataset_name, "data.yaml")

# Tuned Model Path Configuration
model_directory = current_working_directory.parent.parent.parent / "models" / "detection"
#tuned_model_name = "24-04-2026_11-32_yolo26l"
#full_model_weights_path = os.path.join(model_directory , tuned_model_name, "weights", "best.pt")
```

#### Model Setup

```python
# Detection Model
detection_model = YOLO(base_model_name) # Can additionally load from a saved point
```

#### Model Training

```python
# Training Instructions
results = detection_model.train(
    data=str(full_dataset_path),
    project= str(model_directory),
    name = project_name,
    epochs=150,             # more epochs to compensate for small dataset
    imgsz=1280,             # Image Size
    patience=50,            # number of epochs before stopping, if no improvements are seen
    batch=4,                # batch size
    save=True,
    save_period=50,         # checkpoint every 50 epochs
    #optimizer="AdamW",
    #lr0=0.0005,            # lower LR — AdamW works best around here
    #lrf=0.01,              # final LR = lr0 * lrf, gentle decay
    #warmup_epochs=5,       # warm up epochs
    #weight_decay=0.0005,
    classes=[0, 1, 2, 3],
    resume=False,
    plots=True,

    # --- Augmentations ---
    scale=0.5,              # Default
    translate=0.1,          # Default
    fliplr=0.5,             # Default
    flipud=0.0,             # Default
    degrees=5.0,            # Default
    mosaic=1.0,             # Default
    mixup=0.0,              # Default

    # Color
    hsv_h=0.015,            # Default
    hsv_s=0.7,              # Default
    hsv_v=0.4,              # Default
    auto_augment="randaugment"  
)
```

#### Imports

```python
# Imports
import os
import sys
import pandas as pd
```

#### Path Configurations

```python
# Get the absolute path to the 'project' directory
project_root = os.path.abspath(os.path.join('..', '..'))

# Add it to sys.path if it's not already there
if project_root not in sys.path:
    sys.path.append(project_root)

# Path Configuration
video_collection_name = "premier_league_highlights_2026"
data_directory = os.path.join(project_root, "data")
raw_video_collection_directory = os.path.join(data_directory, "raw_videos" , video_collection_name)
os.makedirs(raw_video_collection_directory, exist_ok=True)
```

#### Setting Up Dataframe

```python
df = pd.DataFrame(columns=['video_id', 'video_link', 'video_title'])
```

#### Utils Imports

```python
from utils.video import download_video
```

#### Downloading specified videos

```python
# Specify the videos to download
videos_to_download = [
"https://www.youtube.com/watch?v=W-oRSmyhKO8",
"https://www.youtube.com/watch?v=Ug8imcc-wZg",
"https://www.youtube.com/watch?v=QwwhevIAMd0",
"https://www.youtube.com/watch?v=d8zi8XhQrns&t=3s",
"https://www.youtube.com/watch?v=Th9835EQBlw",
"https://www.youtube.com/watch?v=8RZPMIfcc_g",
"https://www.youtube.com/watch?v=Q3HmXUGm5FA",
"https://www.youtube.com/watch?v=TSJBPED_WEU",
"https://www.youtube.com/watch?v=qOfYEHwj2fM",
"https://www.youtube.com/watch?v=oWkqjOU2uCk"
]

# Walking through files
dataframe_rows = []
for video_link in videos_to_download:
   video_id, title = download_video(video_link, raw_video_collection_directory)
   dataframe_rows.append({'video_id': video_id,
                          'video_link': video_link,
                          'video_title': title
                          })
```

#### Completing and saving Dataframe

```python
df = pd.concat([df, pd.DataFrame(dataframe_rows)], ignore_index=True)
csv_path = os.path.join(raw_video_collection_directory, "collection_info.csv")
df.to_csv(csv_path)
```

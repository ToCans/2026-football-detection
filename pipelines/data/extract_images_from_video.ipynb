#### Imports

```python
# Imports
import os
import sys
```

#### Path Configurations

```python
# Get the absolute path to the 'project' directory
project_root = os.path.abspath(os.path.join('..', '..'))

# Add it to sys.path if it's not already there
if project_root not in sys.path:
    sys.path.append(project_root)

# Path Configuration
data_directory = os.path.join(project_root, "data")
raw_video_data_directory = os.path.join(data_directory, "raw_videos" , "premier_league_highlights_2026")
unannotated_images_output_directory = os.path.join(data_directory, "unannotated_images", "premier_league_highlights_2026", "images")
os.makedirs(unannotated_images_output_directory, exist_ok=True)
```

#### Utils Imports

```python
from utils.video import convert_video_to_images
```

#### Extracting Screenshots

```python
# Walking through files
for root, dirs, files in os.walk(raw_video_data_directory):
    for file in files:
        if file.endswith(".mp4"):
            full_video_file_path = os.path.join(root, file)
            convert_video_to_images(full_video_file_path, 5, unannotated_images_output_directory)
```

#### Imports

```python
# Imports
import os
import sys
import random
import shutil
from pathlib import Path
import numpy as np
```

#### Path Configurations

```python
# Get the absolute path to the 'project' directory
project_root = os.path.abspath(os.path.join('..', '..'))

# Add it to sys.path if it's not already there
if project_root not in sys.path:
    sys.path.append(project_root)

# Path Configuration
class_label = 0
class_multiplier = 5
dataset_name = "football_players_detection"
data_directory = os.path.join(project_root, "data")
detection_dataset_directory_path = os.path.join(data_directory, "detection", dataset_name)
detection_output_directory = os.path.join(data_directory, "detection", f"{dataset_name}_{class_multiplier}x_balls")

# Making necessary directories
os.makedirs(detection_output_directory, exist_ok=True)
```

#### Configuration

```python
PARTITIONS   = ["train", "valid", "test"]   # processed only if they exist
SCALE_RANGE  = (0.7, 1.3)                   # random resize factor for pasted balls
BRIGHTNESS   = 40                           # +/- pixel brightness delta
MAX_ATTEMPTS = 50                           # placement attempts before giving up
SEED         = 42

random.seed(SEED)
np.random.seed(SEED)
```

#### Utility Functions

```python
from utils.data import process_partition
```

#### Dataset Preparation

```python
# ──────────────────────────────────────────────
# Copy dataset metadata (data.yaml, README, etc.)
# ──────────────────────────────────────────────

def copy_metadata(src_root, dst_root):
    for f in Path(src_root).iterdir():
        if f.is_file():
            shutil.copy(f, Path(dst_root) / f.name)


# ──────────────────────────────────────────────
# Run
# ──────────────────────────────────────────────

print("=" * 60)
print("  Ball Augmentation — YOLO Dataset")
print("=" * 60)
print(f"  Source  : {detection_dataset_directory_path}")
print(f"  Output  : {detection_output_directory}")
print(f"  Class ID: {class_label}   |   Copies: {class_multiplier}x per image")
print("=" * 60)

copy_metadata(detection_dataset_directory_path, detection_output_directory)

for partition in PARTITIONS:
    process_partition(
        partition  = partition,
        src_root   = detection_dataset_directory_path,
        dst_root   = detection_output_directory,
        ball_class = class_label,
        num_copies = class_multiplier,
        brighntess=BRIGHTNESS,
        scale=SCALE_RANGE,
        max_attempts=MAX_ATTEMPTS
    )

print("=" * 60)
print(f"  Done! Augmented dataset saved to:")
print(f"  {detection_output_directory}")
print("=" * 60)
```
